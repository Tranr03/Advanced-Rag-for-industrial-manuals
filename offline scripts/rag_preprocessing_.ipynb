## Cell 1: Imports

```python
import sys
sys.path.insert(0, ".")

from rag_preprocessor import RAGPreprocessor
print("Imports OK!")
```

## Cell 2: Ορισμός φακέλων

```python
from config import MANUALS_FOLDER, CHUNKS_FOLDER, IMAGES_FOLDER, MARKDOWN_FOLDER

preprocessor = RAGPreprocessor()
print(f"Manuals folder: {MANUALS_FOLDER}")
```

## Cell 3: Βήμα 1 — Εξαγωγή κειμένου + Chunking

```python
results = preprocessor.process_all(MANUALS_FOLDER)

for filename, (n_text, n_vision) in results.items():
    print(f"{filename}: {n_text} text chunks, {n_vision} vision chunks")
```

## Cell 4: Βήμα 2 — Εξαγωγή Εικόνων με Context

```python
# Το Βήμα 2 εκτελείται πλέον αυτόματα μέσα στο RAGPreprocessor.process_all()
print("Text + Vision indexing ολοκληρώθηκε από το RAGPreprocessor.")
```

## Cell 5: Έλεγχος αποτελεσμάτων

```python
all_meta = preprocessor.collection.get(include=["metadatas"])
sources  = sorted({str(m["source"]) for m in all_meta["metadatas"]})

print("=== ΣΥΝΟΨΗ ===")
print(f"PDFs indexed: {len(sources)}")
for src in sources:
    count = sum(1 for m in all_meta["metadatas"] if m["source"] == src)
    print(f"  📄 {src}: {count} chunks")
print(f"\nΣύνολο documents: {preprocessor.collection.count()}")
```
